```python
a = np.array([3,7,8])
b = np.array([5,7,2])
print('a . b  (dot)   :', np.dot(a, b))
print('a x b  (cross) :', np.cross(a, b))
```

```python
print('L2 norm  ||a||_2 :', la.norm(a))
print('L1 norm  ||a||_1 :', la.norm(a, 1))
print('Linf norm        :', la.norm(a, np.inf))
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))
print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
import numpy as np
def cosine_similarity(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0])),
]
for i, (u, v) in enumerate(pairs, start=1):
    l1 = np.linalg.norm(u, ord=1)
    l2 = np.linalg.norm(u)
    cos = cosine_similarity(u, v)
    print(f"Pair {i}:")
    print("u =", u)
    print("v =", v)
    print("L1 Norm of u:", l1)
    print("L2 Norm of u:", l2)
    print("Cosine Similarity:", cos)
    print()
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])
print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))
print('A symmetric?', np.allclose(A, A.T))
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])
inv_M = np.linalg.inv(M)
trace_M = np.trace(M)
print("Inverse of M:")
print(inv_M)
print("\nTrace of M:", trace_M)
identity = np.eye(2)
print("\nM @ inv(M):")
print(M @ inv_M)
print("\nIs M @ inv(M) equal to Identity?")
print(np.allclose(M @ inv_M, identity))
print("\nIs M symmetric?")
print(np.array_equal(M, M.T))
print("\nIs P orthogonal?")
print(np.allclose(P.T @ P, np.eye(2)))
```

```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])
scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])
fig, ax = plt.subplots(figsize=(4,4))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='pink', lw=0.5); ax.axvline(0, color='pink', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])
def close_loop(points):
    return np.hstack((points, points[:, :1]))
S = np.array([[2,   0],
              [0, 0.5]])
theta = np.radians(45)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
scaled = S @ tri
transformed = R @ scaled
tri_closed = close_loop(tri)
transformed_closed = close_loop(transformed)
plt.figure(figsize=(6,6))
plt.plot(tri_closed[0], tri_closed[1], 'bo-', label="Original")
plt.plot(transformed_closed[0], transformed_closed[1], 'ro-', label="Transformed")
plt.title("Original vs Transformed Triangle")
plt.xlabel("X")
plt.ylabel("Y")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.show()
```

```python
A = np.array([[2., 0.],
              [0., 3.]])
vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
import numpy as np
import numpy.linalg as la

C = np.array([[4., 1.],
              [2., 3.]])
eigenvalues, eigenvectors = la.eig(C)
print("Eigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
print("\nEigenvalues are:")
print(eigenvalues)

v = eigenvectors[:, 0]
lam = eigenvalues[0]

print("\nC @ v:")
print(C @ v)

print("\nlambda * v:")
print(lam * v)

print("\nVerification:")
print(np.allclose(C @ v, lam * v))
```

```python
A2 = np.array([[3., 2.],
               [1., 4.]])
b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

x = np.linalg.solve(A2, b2)

print("Solution x:")
print(x)
rank = np.linalg.matrix_rank(A2)

print("\nRank of A2:")
print(rank)

cat_dog = cosine_similarity(cat, dog)
cat_car = cosine_similarity(cat, car)

print("\nCosine Similarity (cat, dog):", cat_dog)
print("Cosine Similarity (cat, car):", cat_car)

if cat_dog > cat_car:
    print("\ncat and dog are more similar.")
else:
    print("\ncat and car are more similar.")
```
