```python
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
def f(x):
    return x**2
h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
expr=x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
import sympy as sp
def g(x):
    return x**3 + 2*x
h = 1e-6
x0 = 2
numerical_derivative = (g(x0 + h) - g(x0)) / h
print("Numerical Derivative at x = 2:", numerical_derivative)
x = sp.Symbol('x')
g_sym = x**3 + 2*x
symbolic_derivative = sp.diff(g_sym, x)
print("\nSymbolic Derivative:")
print(symbolic_derivative)
symbolic_value = symbolic_derivative.subs(x, 2)

print("\nSymbolic Derivative at x = 2:", symbolic_value)
print("\nDo they match?")
print(abs(numerical_derivative - float(symbolic_value)) < 1e-6)
```

```python
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

```python
x, y = sp.symbols('x y')
h2 = x**2 * y + sp.sin(y)
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)
print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)
gradient = [dh_dx, dh_dy]
print("\nGradient:")
print(gradient)
gradient_at_point = [
    dh_dx.subs({x: 2, y: 0}),
    dh_dy.subs({x: 2, y: 0})
]
print("\nGradient at (2, 0):")
print(gradient_at_point)
```

```python
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)
print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))
```

```python
x = sp.Symbol('x')
y = sp.exp(x**2 + 1)
by_hand = sp.exp(x**2 + 1) * (2 * x)
print("By Hand Derivative:")
print(by_hand)
by_sympy = sp.diff(y, x)
print("\nSymPy Derivative:")
print(by_sympy)
print("\nDo they match?")
print(sp.simplify(by_hand - by_sympy) == 0)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1
z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
import numpy as np

h  = np.maximum(0, z1)
y_hat = h @ W2
dy   = 2 * (y_hat - Y) / Y.size      # d loss / d y_hat
dW2  = h.T @ dy                      # d loss / d W2
dh   = dy @ W2.T                     # d loss / d h
dz1  = dh * (z1 > 0)                 # ReLU gradient (1 where z1>0 else 0)
dW1  = X.T @ dz1                     # d loss / d W1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
import numpy as np

h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
import numpy as np
np.random.seed(0)
Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1
Wb = np.random.randn(8, 1) * 0.1
z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb
loss = np.mean((y_hat - Yb) ** 2)
print("Loss before update:", loss)
dy = (2 / len(Xb)) * (y_hat - Yb)
dWb = h.T @ dy
dh = dy @ Wb.T
dz1 = dh * (z1 > 0)
dWa = Xb.T @ dz1
lr = 0.05
Wa = Wa - lr * dWa
Wb = Wb - lr * dWb
z1_new = Xb @ Wa
h_new = np.maximum(0, z1_new)
y_hat_new = h_new @ Wb
loss_new = np.mean((y_hat_new - Yb) ** 2)

print("Loss after update:", loss_new)
```

```python
xv = 0.0
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)
    xv = xv - lr * grad
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

```python
import sympy as sp
x, y = sp.symbols('x y')
f = x**4 + y**2
H = sp.hessian(f, (x, y))
print("Hessian Matrix:")
sp.pprint(H)
xv = 0.0
lr = 0.1
for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad
print("\nFinal x:", xv)
```
